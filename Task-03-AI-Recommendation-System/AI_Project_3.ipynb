```python
# ==========================================
# AI Recommendation System
# Project 3 - DecodeLabs
# ==========================================

print("==========================================")
print("     AI Recommendation System")
print("==========================================")
print("Welcome! 😊")
print("This system recommends courses based on your interests.")
print()
```

```python
# Step 2: Taking user interests

interest1 = input("Enter your first interest: ")
interest2 = input("Enter your second interest: ")
interest3 = input("Enter your third interest: ")

user_interests = [interest1, interest2, interest3]

print()
print("Your interests are:")
print(user_interests)
```

```python
# Step 3: Creating a dataset

courses = {
    "Python for Beginners": ["python", "programming", "coding"],
    "Machine Learning": ["python", "ai", "machine learning"],
    "Artificial Intelligence": ["ai", "machine learning", "deep learning"],
    "Web Development": ["html", "css", "javascript"],
    "Data Science": ["python", "data", "machine learning"],
    "Graphic Design": ["photoshop", "illustrator", "design"],
    "Flutter App Development": ["flutter", "dart", "mobile"],
    "Cyber Security": ["network", "security", "hacking"]
}

print("Available Courses:\n")

for course in courses:
    print(course)
```

```python
# Step 4: Matching user interests with course tags

print("\nRecommended Courses:\n")

for course, tags in courses.items():

    match_count = 0

    for interest in user_interests:

        if interest.lower() in tags:
            match_count += 1

    if match_count > 0:
        print(course, "->", match_count, "interest(s) matched")
```

```python
# Step 5: Sorting recommendations by highest score

recommendations = []

for course, tags in courses.items():
    match_count = 0

    for interest in user_interests:
        if interest.lower() in tags:
            match_count += 1

    if match_count > 0:
        recommendations.append((course, match_count))

recommendations.sort(key=lambda x: x[1], reverse=True)

print("Top Recommended Courses:\n")

for course, score in recommendations:
    print(course, "-> Score:", score)
```

```python
print("========== TOP 3 RECOMMENDATIONS ==========\n")

top_3 = recommendations[:3]

for i, (course, score) in enumerate(top_3, start=1):
    print(f"{i}. {course} (Score: {score})")
```

```python
# Step 7: Display Best Recommendation

print("\n================================")

if len(recommendations) > 0:
    best_course = recommendations[0]

    print("🏆 Best Recommendation")
    print("Course :", best_course[0])
    print("Score  :", best_course[1])

else:
    print("No recommendation found.")

print("================================")
```

```python
# Step 8: Display Match Percentage

print("\n========== MATCH PERCENTAGE ==========\n")

for course, score in recommendations:
    percentage = (score / len(user_interests)) * 100
    print(f"{course} --> {percentage:.0f}% Match")
```

```python
# Step 3: Taking user preferences for AI recommendation

user_input = input("Enter your interests separated by commas: ")

print("User Profile:")
print(user_input)
```

```python
# Step 4: Convert text into vectors using TF-IDF

from sklearn.feature_extraction.text import TfidfVectorizer

# Ensure course_features is defined, in case previous cells were not run or kernel restarted.
course_names = [
    "Python for Beginners",
    "Machine Learning",
    "Artificial Intelligence",
    "Web Development",
    "Data Science",
    "Graphic Design",
    "Flutter App Development",
    "Cyber Security"
]

course_features = [
    "python programming coding beginner",
    "python ai machine learning algorithms",
    "artificial intelligence ai machine learning deep learning",
    "html css javascript frontend website",
    "python data analysis machine learning statistics",
    "photoshop illustrator design creativity",
    "flutter dart mobile app development",
    "network security hacking cyber protection"
]


all_text = course_features + [user_input]

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(all_text)

print("Text converted into vectors successfully!")
print("Matrix Shape:", tfidf_matrix.shape)
```

```python
# Step 5: Calculate Cosine Similarity

from sklearn.metrics.pairwise import cosine_similarity

similarity_scores = cosine_similarity(
    tfidf_matrix[-1:],      # User Input
    tfidf_matrix[:-1]       # All Courses
)

print("Similarity Scores:\n")
print(similarity_scores)
```

```python
# Step 6: Attach course names with similarity scores

scores = similarity_scores[0]

ai_recommendations = []

for i in range(len(course_names)):
    ai_recommendations.append((course_names[i], scores[i]))

ai_recommendations.sort(key=lambda x: x[1], reverse=True)

print("===== AI Recommended Courses =====\n")

for course, score in ai_recommendations[:3]:
    print(f"{course} --> {score:.2f} similarity")
```

```python
print("\n🏆 Best AI Recommendation\n")

best_course = ai_recommendations[0]

print("Course:", best_course[0])
print("Similarity Score:", round(best_course[1], 2))

print("\nThank you for using the AI Recommendation System!")
```
